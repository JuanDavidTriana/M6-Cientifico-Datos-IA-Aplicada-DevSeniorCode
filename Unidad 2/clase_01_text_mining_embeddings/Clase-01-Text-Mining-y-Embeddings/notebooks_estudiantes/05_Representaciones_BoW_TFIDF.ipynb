## Objetivo
Recorrer la **evolución de las representaciones de texto** antes de los embeddings, entender sus ventajas y, sobre todo, sus **limitaciones** (que motivan el notebook 06).

| Técnica | Idea | Limitación principal |
|---|---|---|
| One-Hot | Un 1 en la posición de la palabra | Vectores gigantes, sin semántica |
| Bag of Words | Conteo de palabras por documento | Pierde orden y contexto |
| TF-IDF | Conteo ponderado por rareza | Sigue sin entender sinónimos |
| Embeddings | Vector denso aprendido | Requiere datos/entrenamiento |

## 3) TF-IDF
**TF** (frecuencia del término en el documento) × **IDF** (qué tan raro es en el corpus).

$$\text{tfidf}(t,d) = \text{tf}(t,d) \cdot \log\frac{1+N}{1+\text{df}(t)} + 1$$

Palabras que aparecen en todos los documentos (como `el`) pesan **menos**; palabras distintivas pesan **más**.
