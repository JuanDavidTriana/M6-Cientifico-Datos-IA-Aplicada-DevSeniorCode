## 6) ¿Cuándo usar `pre` o `post`?
- `post`: intuitivo, conserva el inicio; es lo habitual con `Embedding(mask_zero=True)` y en arquitecturas modernas.
- `pre`: clásico en RNN/LSTM antiguas, donde la información final queda "más cerca" de la salida.

> 💡 **Máscara**: con `Embedding(..., mask_zero=True)` el modelo ignora los ceros del padding. Lo usaremos en el notebook 07.
